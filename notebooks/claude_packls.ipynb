```python
# What is actually inside the three pilkwang datasets every 0.94x notebook mounts?
# Every public config points at weights/unet_transformer/split_0/edge_predictor_best.pth.
# If the pack ships more splits or more checkpoints, that is an axis nobody is using.
from pathlib import Path
for root in sorted(Path("/kaggle/input").glob("*")):
    print("=" * 78)
    print(root.name)
    files = sorted(p for p in root.rglob("*") if p.is_file())
    print(f"  {len(files)} files")
    from collections import Counter
    ext = Counter(p.suffix for p in files)
    print("  ext:", dict(ext.most_common(8)))
    for p in files:
        if p.suffix in (".pt", ".pth", ".ckpt", ".json", ".yaml", ".yml", ".txt"):
            print(f"    {p.stat().st_size:>13,}  {p.relative_to(root)}")
```
