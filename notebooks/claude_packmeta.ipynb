```python
# The model authors' own documentation, which nobody in the fork lineage quotes.
# Of particular interest: what distinguishes checkpoint_last from best (the 0.948 branch
# swaps exactly that for deepcenter), and gate_summary.json, which may be where
# DEEPCENTER_SAFE_DIV_THRESHOLD's values came from.
import json
from pathlib import Path
WANT = ("ARTIFACT_MANIFEST.json", "SNAPSHOT_MANIFEST.json", "config.json",
        "training_config.json", "gate_summary.json")
for p in sorted(Path("/kaggle/input").rglob("*.json")):
    if p.name not in WANT:
        continue
    print("=" * 78)
    print(p.relative_to("/kaggle/input"), f"({p.stat().st_size:,} bytes)")
    try:
        d = json.loads(p.read_text())
    except Exception as e:
        print("  unreadable:", e); continue
    s = json.dumps(d, indent=1)
    print(s[:3500])
```

```python
# split_manifest.json separately: notes/24 called split_0 membership unknowable.
import json
from pathlib import Path
for p in sorted(Path("/kaggle/input").rglob("split_manifest.json")):
    d = json.loads(p.read_text())
    print("=" * 78); print(p.relative_to("/kaggle/input"))
    for k, v in d.items():
        if isinstance(v, list):
            print(f"  {k}: {len(v)} entries -> {v[:4]}{' ...' if len(v) > 4 else ''}")
        else:
            print(f"  {k}: {v}")
    PLACEHOLDERS = ["44b6_0113de3b", "44b6_0b24845f", "6bba_05b6850b", "6bba_05db0fb1"]
    flat = json.dumps(d)
    print("  placeholder clips present in this manifest:",
          [c for c in PLACEHOLDERS if c in flat])
```
