```python
# gate_threshold_metrics.csv is a THRESHOLD SWEEP the model's author ran and published
# inside the artifact. BIOHUB_DEEPCENTER_SAFE_DIV_THRESHOLD (0.12 -> 0.25 was half of the
# public 0.939 -> 0.941 step) and BIOHUB_DEEPCENTER_GAP_THRESHOLD (0.25) are exactly this
# gate. If the sweep names an optimum, the arm `dc40` can be aimed instead of guessed.
# history.csv says which epoch best.pt and checkpoint_last.pt are, which is what separates
# the claimed-0.948 branch from the 0.941 line.
from pathlib import Path
for name in ("gate_threshold_metrics.csv", "gate_summary.json"):
    for p in sorted(Path("/kaggle/input").rglob(name)):
        print("=" * 78); print(p.relative_to("/kaggle/input"))
        print(p.read_text()[:3000])
```

```python
from pathlib import Path
for p in sorted(Path("/kaggle/input").rglob("history.csv")):
    print("=" * 78); print(p.relative_to("/kaggle/input"))
    rows = p.read_text().splitlines()
    print(f"{len(rows)-1} rows")
    print(rows[0])
    for r in rows[1:6]: print(" ", r)
    print("  ...")
    for r in rows[-6:]: print(" ", r)
    # which epoch is "best"? report the row minimising each loss-like column
    hdr = rows[0].split(",")
    import math
    data = [r.split(",") for r in rows[1:] if r.strip()]
    for j, h in enumerate(hdr):
        vals = []
        for r in data:
            try: vals.append((float(r[j]), r[0]))
            except Exception: pass
        if len(vals) == len(data) and len(vals) > 3 and h.lower() not in ("epoch", "step"):
            lo = min(vals); hi = max(vals)
            print(f"   {h:<26} min {lo[0]:.5f} @ {lo[1]}   max {hi[0]:.5f} @ {hi[1]}   last {vals[-1][0]:.5f}")
```
